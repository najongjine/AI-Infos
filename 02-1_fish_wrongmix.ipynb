**아래는 실무자가 GPT와 짠 코드. 차원이 틀림**


```python
import numpy as np
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
import matplotlib.pyplot as plt

# 1. 데이터 만들기
fish_data = np.column_stack((fish_length, fish_weight))  # shape: (49, 2)
fish_target = np.concatenate((np.ones(35), np.zeros(14)))  # 도미:1, 빙어:0
print(f"fish_data : {fish_data}")
print(f"fish_target : {fish_target}")

# 2. 합치기 (데이터 + 정답 라벨)
full_data = np.column_stack((fish_data, fish_target))  # shape: (49, 3)


# 3. 섞기
np.random.seed(42)
np.random.shuffle(full_data)  # 행 기준 섞임
print(f"full_data: {full_data}")

# 4. 다시 나누기
"""
: → 모든 행 (49개 생선 전부)
:2 → 열 0부터 2 전까지 = 열 0, 열 1
"""
shuffled_input = full_data[:, :2]  # 앞 2개 열 = 길이 + 무게
"""
: → 모든 행 (49개 생선 전부)
:2 → 열 번호 2 = 정답 라벨만 뽑기
"""
shuffled_target = full_data[:, 2]  # 마지막 열 = 정답

# 5. 훈련/테스트 분리
train_input = shuffled_input[:35]
train_target = shuffled_target[:35]
test_input = shuffled_input[35:]
test_target = shuffled_target[35:]
```

```python
kn.fit(train_input, train_target)
# 7. 테스트 데이터로 예측
predict_result = kn.predict(test_input)
print("예측 결과:", predict_result)
print("실제 정답 :", test_target)

# 8. 정확도 평가
score = kn.score(test_input, test_target)
print(f"정확도: {score:.2f}")

# 9. 시각화 (선택)
plt.scatter(train_input[:, 0], train_input[:, 1],  marker='o')  # 고정 색
plt.scatter(test_input[:, 0], test_input[:, 1],  marker='s')   # 고정 색

plt.xlabel('Length')
plt.ylabel('Weight')
plt.legend()
plt.title("Fish Classification")
plt.grid(True)
plt.show()
```

```python
# 유저 입력 예시
user_fish = np.array([[30, 600]])
prediction = kn.predict(user_fish)
print("예측 결과:", prediction)
print("이 생선은", "도미" if prediction[0] == 1 else "빙어", "입니다.")
```
